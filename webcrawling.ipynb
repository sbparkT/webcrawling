````python
import requests
from bs4 import BeautifulSoup
````

````python
#1. url 요청,웹페이지 가져오기
res=requests.get('https://www.marieclairekorea.com/fashion/2026/05/ai-glass/')
````

````python
#2.웹페이지 파싱하기(문자열 의미 분석>)
soup=BeautifulSoup(res.content,'html.parser')
soup
````

[html구조이해](https://wikidocs.net/160892)

```
find(tag,attrs,recursive.text,limit,**kwargs)  1개의 데이터 가져오기
find_all(tag,attrs,recursive.text,limit,**kwargs) 찾고자 하는 모든 데이터 가져오기
 *attrs = "이 태그 중에서 이 속성을 가진 것만"
get_text() : 문자로 가져오기
```



````python
#3. 필요 데이터 추출
mydata=soup.find('title') #개발자 도구 활용해 필요tag부분 추출해보기
mydata
````

````python
print(mydata.get_text())
````

````python
mydata=soup.find('h2')
print(mydata.get_text()) #가장 처음 것 추출
````

````python
mydata2=soup.find_all('h2')
print(mydata2) #모든 것 추출 ->get_text는 단일 요소에만 사용가능 .반복문으로 개별적으로 가져와
for h2_tag in mydata2:
    print(h2_tag.get_text())
````

````python
data=soup.find_all('div')
print(data)
````

````python
data=soup.find_all('div',{'class':{'post-content'}})
print(data)
````

````python
all_p_tags = soup.find_all('p')
for p_tag in all_p_tags:
  print(p_tag.get_text())
print(all_p_tags[1].get_text())
````

````python
#본문 글 가져오기
all_p_tags = soup.find_all('p')
for p_tag in all_p_tags:
  # 'style' 속성을 확인하여 특정 스타일(회사 정보)을 가진 태그는 skip
  style_attr = p_tag.get('style')
  print(style_attr)
  if style_attr and 'font-size: 12px;' in style_attr:
    continue
  print(p_tag.get_text())
````

````python
#멜론차트 가져오기
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0.0.0 Safari/537.36"
}
url="https://www.melon.com/chart/index.htm"
response = requests.get(url, headers=headers)
soup=BeautifulSoup(response.content,'html.parser')
````

````python
#html도 결국 트리구조 : 자식 태그 확인
for child in soup.find('div',{'id':'tb_list'}).children:
  print(child)
````

````python
#html도 결국 트리구조 : 형제 태그 확인
for sibling in soup.find('th',{'scope':'col'}).next_siblings:
  print(sibling)
````
